```python
import sqlite3
import pandas as pd
```

```python
#change the query (date part) to select data respectively for train and test sets
query = """
SELECT 
    month,
    day_of_week,
    is_promo,
    is_holiday,
    amount
FROM sales
WHERE date >= '2025-01-31'
"""

with sqlite3.connect('data.db') as conn:
    df = pd.read_sql_query(query, conn)

print(df.head())
print(df.shape)

```

```python
df.to_csv('train.csv', index=False) 
```

```python
df.to_csv('test.csv', index=False) 
```
